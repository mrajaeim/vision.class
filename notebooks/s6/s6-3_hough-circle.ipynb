```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
smarties_img = cv.imread("../../../class.vision/images/smarties.png")
old_iranian_coins_img = cv.imread("../../../class.vision/images/old-iranian-coins.jpg")
iranian_coin_img = cv.imread("../../../class.vision/images/iranian-coin.jpg")
```

```python
def to_gray(img):
    return cv.cvtColor(img, cv.COLOR_BGR2GRAY)
```

```python
w = smarties_img.shape[1]
draw_img = smarties_img.copy()
blured = cv.medianBlur(smarties_img, 5)
circles = cv.HoughCircles(to_gray(blured), cv.HOUGH_GRADIENT, 1, w/8,param1=100,param2=30,minRadius=0,maxRadius=100)
circles = np.round(circles[0]).astype('uint16')
for circle in circles:
    cv.circle(draw_img, (circle[0], circle[1]), circle[2], (0,255,200), 3)


plt.imshow(draw_img[...,::-1])
```

```python
w = old_iranian_coins_img.shape[1]
draw_img = old_iranian_coins_img.copy()
blured = cv.medianBlur(old_iranian_coins_img, 5)
circles = cv.HoughCircles(to_gray(blured), cv.HOUGH_GRADIENT, 1, w/8,param1=100,param2=30,minRadius=0,maxRadius=100)
circles = np.round(circles[0]).astype('uint16')
for circle in circles:
    cv.circle(draw_img, (circle[0], circle[1]), 5, (0,0,255), -1)
    cv.circle(draw_img, (circle[0], circle[1]), circle[2], (0,255,0), 3)


plt.imshow(draw_img[...,::-1])
```

```python
w = iranian_coin_img.shape[1]
draw_img = iranian_coin_img.copy()
blured = cv.medianBlur(iranian_coin_img, 5)
circles = cv.HoughCircles(to_gray(blured), cv.HOUGH_GRADIENT, 1, w/10,param1=100,param2=50,minRadius=50,maxRadius=100)
circles = np.round(circles[0]).astype('uint16')
for circle in circles:
    cv.circle(draw_img, (circle[0], circle[1]), 5, (0,0,255), -1)
    cv.circle(draw_img, (circle[0], circle[1]), circle[2], (0,255,0), 3)


plt.imshow(draw_img[...,::-1])
```
