```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv.imread("../../../class.vision/images/Lincoln.jpg")
mask = cv.imread("../../../class.vision/images/inpaintMask.png",0)


res_telea = cv.inpaint(image, mask, 3, cv.INPAINT_TELEA)
res_ns = cv.inpaint(image, mask, 3, cv.INPAINT_NS)

plt.figure(figsize=[18,7])
plt.subplot(141);plt.imshow(image[...,::-1]);plt.title("image");
plt.subplot(142);plt.imshow(mask, cmap='gray');plt.title("mask");
plt.subplot(143);plt.imshow(res_telea[...,::-1]);plt.title("TELEA");
plt.subplot(144);plt.imshow(res_ns[...,::-1]);plt.title("NS");
```
