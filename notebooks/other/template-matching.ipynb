```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
cow_img = cv.imread('../../../class.vision/images/input.jpg')
```

```python
def to_gray(img):
    return cv.cvtColor(img, cv.COLOR_BGR2GRAY)
```

```python
template = to_gray(cow_img[280:400, 300:520,::-1])
src_img = to_gray(cow_img)
res = cv.matchTemplate(src_img, template, cv.TM_CCORR_NORMED)
h, w = template.shape[:2]
minVal, maxVal, minLoc, maxLoc = cv.minMaxLoc(res)
x, y = maxLoc
roi = cow_img[y:y+h, x:x+w]

plt.figure(figsize=[12,6])
plt.subplot(121);plt.imshow(res, cmap='viridis'); plt.colorbar()
plt.subplot(122);plt.imshow(roi[...,::-1]); plt.title("ROI");
```
