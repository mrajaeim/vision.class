```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
END = "\n" + "-" * 50 + "\n"
```

```python
a = np.array([f"a{i}" for i in range(4)])
b = np.array([f"b{i}" for i in range(4)])
c = np.array([f"c{i}" for i in range(4)])
d = np.array([f"d{i}" for i in range(4)])
```

## reshape

```python
print(np.array(a), end=END)
print(np.array(a).reshape(-1,1), end=END)
print(np.array(a).reshape(-1,2), end=END)
print(np.array(a).reshape(-1,4), end=END)
```

## column_stack

```python
np.column_stack((a,b,c,d))
```

```python
np.column_stack((a,b))
```

```python
np.column_stack((a.reshape(-1, 1),b.reshape(-1, 1)))
```

## hstack

```python
np.hstack((a,b))
```

```python
np.hstack((a.reshape(-1, 1),b.reshape(-1, 1)))
```
