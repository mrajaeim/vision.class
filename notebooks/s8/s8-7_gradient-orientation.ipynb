```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
def grad_orien(path):
    img = cv.imread(path, 0)
    sx = cv.Sobel(img, cv.CV_32F, 1, 0, ksize=3)
    sy = cv.Sobel(img, cv.CV_32F, 0, 1, ksize=3)

    plt.figure(figsize=(10,6))
    plt.subplot(131); plt.imshow(img, cmap='gray'); plt.title("")
    plt.subplot(132); plt.imshow(sx, cmap='gray'); plt.title("sx")
    plt.subplot(133); plt.imshow(sy, cmap='gray'); plt.title("sy")
    plt.show()

    res = cv.phase(np.float32(sx), np.float32(sy), angleInDegrees=True)
    plt.figure(figsize=(20,5))
    plt.subplot(131); plt.imshow(res, cmap="tab10"); plt.title(""); plt.colorbar(ticks=np.arange(0, 360, 45))
    plt.tight_layout()

grad_orien("../../images/clock.jpg")
```
