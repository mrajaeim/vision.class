```python
import cv2 as cv
import numpy as np
```

```python
from scipy import io

dataset = io.loadmat("../../../class.vision/dataset/Data_hoda_full.mat")
train_X = dataset['Data'][:1000].squeeze()
train_Y = dataset['labels'][:1000].squeeze().astype(np.int64)
test_X =  dataset['Data'][1000:1200].squeeze()
test_Y =  dataset['labels'][1000:1200].squeeze().astype(np.int64)

train_X = np.array([cv.resize(x, (5,5)) for x in train_X])
test_X = np.array([cv.resize(x, (5,5)) for x in test_X])

train_X = train_X.reshape(-1, 25).astype(np.float32)
test_X = test_X.reshape(-1, 25).astype(np.float32)

print("Train X: ", train_X.shape, "Y: ", train_Y.shape)
print("Test X: ", test_X.shape, "Y: ", test_Y.shape)
```

```python
knn = cv.ml.KNearest_create()
knn.train(train_X, cv.ml.ROW_SAMPLE, train_Y)
```

```python
k = 3
ret, result, neighbors, dist = knn.findNearest(test_X.astype('float32'), k)
```

```python
matches = np.equal(result.ravel(), test_Y)
np.mean(matches)
```
