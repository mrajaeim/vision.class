```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
from scipy import io
```

```python
dataset = io.loadmat("../../../class.vision/dataset/Data_hoda_full.mat")
```

```python
train_X = dataset['Data'][:1000].squeeze()
train_Y = dataset['labels'][:1000].squeeze()
test_X =  dataset['Data'][1000:1200].squeeze()
test_Y =  dataset['labels'][1000:1200].squeeze()

print(train_X.shape)
print(train_Y.shape)
```

```python
i = 90
viz1 = train_X[i], train_Y[i]
viz2 = test_X[i], test_Y[i]

plt.figure(figsize=(10,6))
plt.subplot(121); plt.imshow(viz1[0], cmap='gray'); plt.title(viz1[1])
plt.subplot(122); plt.imshow(viz2[0], cmap='gray'); plt.title(viz2[1])
```

```python
train_X = np.array([cv.resize(x, (5,5)) for x in train_X])
test_X = np.array([cv.resize(x, (5,5)) for x in test_X])
```

```python
i = 90
viz1 = train_X[i], train_Y[i]
viz2 = test_X[i], test_Y[i]

plt.figure(figsize=(10,6))
plt.subplot(121); plt.imshow(viz1[0], cmap='gray'); plt.title(viz1[1])
plt.subplot(122); plt.imshow(viz2[0], cmap='gray'); plt.title(viz2[1])
```

```python
print(train_X.shape)
print(test_X.shape)
train_X = train_X.reshape(-1, 25)
test_X = test_X.reshape(-1, 25)
print(train_X.shape)
print(test_X.shape)
```
