```python
import cv2 as cv
import numpy as np
```

```python
dataset = np.loadtxt('../../../class.vision/dataset/letter-recognition.data', np.float32,
                     delimiter=',',
                     converters={0: lambda ch: ord(ch) - ord('A')}
                     )

print(dataset.shape)
print(dataset[0])
```

```python
train, test = np.vsplit(dataset,2)
print(train.shape)
print(test.shape)
```

```python
Y_train, X_train = np.hsplit(train,[1])
Y_test, X_test = np.hsplit(test,[1])
print(X_train.shape)
print(Y_train.shape)
```

```python
knn = cv.ml.KNearest_create()
knn.train(X_train, cv.ml.ROW_SAMPLE, Y_train)
```

```python
all_acc = []
for i in range(2, 10):
    _, res, _, _= knn.findNearest(X_test, k=i)
    acc = np.mean(np.equal(res, Y_test))
    all_acc.append([i, acc])
    print(f"{i:}\t{acc*100:0.5f}%")
all_acc.sort(key=lambda x: x[1], reverse=True)
print(f"best k: {all_acc[0][0]} \t{all_acc[0][1]*100:0.5f}%")
```
