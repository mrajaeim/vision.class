```python
import cv2 as cv
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
from scipy import io
```

```python
dataset = io.loadmat("../../../class.vision/dataset/Data_hoda_full.mat")
train_X = dataset['Data'][:1000].squeeze()
train_Y = dataset['labels'][:1000].squeeze()
test_X =  dataset['Data'][1000:1200].squeeze()
test_Y =  dataset['labels'][1000:1200].squeeze()

train_X = np.array([cv.resize(x, (5,5)) for x in train_X])
test_X = np.array([cv.resize(x, (5,5)) for x in test_X])

train_X = train_X.reshape(-1, 25)
test_X = test_X.reshape(-1, 25)

print("Train X: ", train_X.shape, "Y: ", train_Y.shape)
print("Test X: ", test_X.shape, "Y: ", test_Y.shape)
```

```python
nc = KNeighborsClassifier(n_neighbors=3)
nc.fit(train_X, train_Y)
```

```python
i = 11
data = [test_X[i]]
[pred] = nc.predict(data)
print ("Sample {} is a {}, and you prediction is: {}.".format(i, test_Y[i], pred))
print(nc.predict_proba(data))
```

```python
pred_classes = nc.predict(test_X)
np.mean(test_Y == pred_classes)
```

```python
acc = nc.score(test_X, test_Y)
print ("Accuracy is %.2f %%" %(acc*100))
```
