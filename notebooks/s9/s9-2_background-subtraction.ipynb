```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
bsub = cv.createBackgroundSubtractorMOG2()
# bsub = cv.createBackgroundSubtractorKNN()

cap = cv.VideoCapture("../../../class.vision/videos/vtest.avi")
kernel = np.ones((5,5),np.uint8)

while 1:
    ret, frame = cap.read()
    if not ret:
        break
    fgmask = bsub.apply(frame)
    fgmask = cv.morphologyEx(fgmask, cv.MORPH_OPEN, kernel, iterations=2, )
    # Write frame index
    cv.rectangle(fgmask, (0,0) , (70, 20), (255,255,255), -1)
    cv.putText(fgmask, str(cap.get(cv.CAP_PROP_POS_FRAMES)), (15, 15), cv.FONT_HERSHEY_SIMPLEX, 0.5, (0,0,0))
    cv.imshow("frame mask",fgmask)
    if cv.waitKey(5) & 0xFF == 27:
        break

cap.release()
cv.destroyAllWindows()
```
