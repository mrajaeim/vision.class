```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
blox_img = cv.imread('../../../class.vision/images/blox.jpg')
```

```python
def to_gray(img):
    return cv.cvtColor(img, cv.COLOR_BGR2GRAY)
```

```python
draw_img = blox_img.copy()
corners = cv.goodFeaturesToTrack(to_gray(blox_img).astype(np.float32),24,0.01,10)
corners = corners.astype(np.int32).ravel().reshape(-1,2)
for corner in corners:
    cv.circle(draw_img, (corner[0], corner[1]), 5, (0,0,255), -1)

plt.figure(figsize=(10,6))
plt.subplot(133); plt.imshow(draw_img[...,::-1]); plt.title("Result")
plt.axis('off');
```
