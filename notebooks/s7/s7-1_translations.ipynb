```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv.imread("../../images/sheep.jpg")
H, W = img.shape[:2]

M = np.array([
    [1, 0, -50], # X
    [0, 1, 200], # Y
]).astype(np.float32)

output = cv.warpAffine(img, M, (W, H))

plt.figure(figsize=(10,6))
plt.subplot(121); plt.imshow(img[...,::-1])
plt.subplot(122); plt.imshow(output[...,::-1])
```
