```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv.imread("../../images/sheep.jpg")
```

```python
low_sigma = cv.GaussianBlur(img,(5,5),2)
high_sigma = cv.GaussianBlur(img,(5,5),7)

dog = (low_sigma - high_sigma)
rev_dog = (high_sigma - low_sigma)

plt.figure(figsize=[20,5])
plt.subplot(151),plt.imshow(img[...,::-1]),plt.title('original');
plt.subplot(152),plt.imshow(low_sigma[...,::-1]),plt.title('Low Sigma Gaussian');
plt.subplot(153),plt.imshow(high_sigma[...,::-1]),plt.title('High Sigma Gaussian');
plt.subplot(154),plt.imshow(dog[...,::-1]),plt.title('DOG');
plt.subplot(155),plt.imshow(rev_dog[...,::-1]),plt.title('Rev DOG');
```
