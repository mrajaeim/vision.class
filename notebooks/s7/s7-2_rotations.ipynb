```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv.imread("../../images/sheep.jpg")
```

## Custom theta

```python
theta = np.deg2rad(10)

M = np.float32([
    [np.cos(theta), -np.sin(theta),0],
    [np.sin(theta),np.cos(theta),0]
])

output = cv.warpAffine(img,M,img.shape[:2][::-1])

plt.figure(figsize=(10,6))
plt.subplot(121); plt.imshow(img[...,::-1])
plt.subplot(122); plt.imshow(output[...,::-1])
```

## Rotation with Translation

```python
theta = np.deg2rad(10)

M = np.float32([
    [np.cos(theta), -np.sin(theta), 50],
    [np.sin(theta),np.cos(theta), 100]
])

output = cv.warpAffine(img,M,img.shape[:2][::-1])

plt.figure(figsize=(10,6))
plt.subplot(121); plt.imshow(img[...,::-1])
plt.subplot(122); plt.imshow(output[...,::-1])
```

## Use getRotationMatrix2D

```python
H, W = img.shape[:2]

M = cv.getRotationMatrix2D((W/2, H/2), 10, 0.8)
print(M)
output = cv.warpAffine(img,M,(W,H))

plt.figure(figsize=(10,6))
plt.subplot(121); plt.imshow(img[...,::-1])
plt.subplot(122); plt.imshow(output[...,::-1])
```

## Flip and Transpose

```python
plt.figure(figsize=(10,6))
plt.subplot(221); plt.imshow(img[...,::-1]); plt.title("Original")
plt.subplot(222); plt.imshow(cv.transpose(img)[...,::-1]); plt.title("Transpose")
plt.subplot(223); plt.imshow(cv.flip(img, 0)[...,::-1]); plt.title("Flip 0")
plt.subplot(224); plt.imshow(cv.flip(img, 1)[...,::-1]); plt.title("Flip 1")
```
