```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv.imread("../../images/sheep.jpg")
```

## Scale Matrix

```python
M = np.float32([
    [0.5, 0,0],
    [0,0.5,0]
])

output = cv.warpAffine(img,M,img.shape[:2][::-1])

plt.figure(figsize=(10,6))
plt.subplot(121); plt.imshow(img[...,::-1])
plt.subplot(122); plt.imshow(output[...,::-1])
```

## Composition of Matrices

```python
theta = np.deg2rad(10)

M1 = np.float32([
    [np.cos(theta), -np.sin(theta)],
    [np.sin(theta),np.cos(theta)]
])

M2 = np.float32([
    [0.5, 0],
    [0,0.5]
])

M12 = np.matmul(M1,M2)

M = np.ones((2,3), dtype=np.float32) * 250
M[:,:-1] = M12

output = cv.warpAffine(img,M,img.shape[:2][::-1])

plt.figure(figsize=(10,6))
plt.subplot(121); plt.imshow(img[...,::-1])
plt.subplot(122); plt.imshow(output[...,::-1])
```

## Resize function

```python
plt.figure(figsize=(10,6))
plt.subplot(321); plt.imshow(img[...,::-1])
plt.subplot(323); plt.imshow(cv.resize(img, None, fx=2, fy=1, interpolation=cv.INTER_NEAREST)[...,::-1])
plt.subplot(324); plt.imshow(cv.resize(img, None, fx=1, fy=2, interpolation=cv.INTER_CUBIC)[...,::-1])
plt.subplot(325); plt.imshow(cv.resize(img, None, fx=3, fy=3, interpolation=cv.INTER_AREA)[...,::-1])
plt.subplot(326); plt.imshow(cv.resize(img, (400,400), cv.INTER_NEAREST)[...,::-1])
```
