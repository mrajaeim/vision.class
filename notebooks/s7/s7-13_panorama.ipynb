```python
import cv2 as cv
import matplotlib.pyplot as plt
from glob import glob
```

```python
imagefiles = glob("../../../class.vision/images/boat*.jpg")
imgs = [cv.imread(imagefile) for imagefile in imagefiles]
stitcher = cv.Stitcher_create()
status, result = stitcher.stitch(imgs)
if status == 0:
    plt.figure(figsize=[30,10])
    plt.imshow(result[...,::-1])
```
