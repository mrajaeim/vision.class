```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
src = cv.imread("../../../class.vision/images/little_girl.jpg")

src = cv.GaussianBlur(src, (3, 3), 0)

edges = cv.Laplacian(src, -1, ksize=3)

edges = cv.convertScaleAbs(edges)

plt.figure(figsize=[12,7])
plt.subplot(121);plt.imshow(src[...,::-1]);plt.title("Original");
plt.subplot(122);plt.imshow(edges, cmap='gray');plt.title("Laplacian");
```
