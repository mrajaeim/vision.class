```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
input_img = cv.imread("../../../class.vision/images/PopEffects.jpeg")
```

```python
window_name = "Pop Effects"

drawing_img = input_img.copy()

def mouse_click(event, x, y, flags, param):
    if event == cv.EVENT_LBUTTONDOWN:
        pixel = input_img[y:y+1, x:x+1, :]
        cv.circle(drawing_img, (x, y), 5, (0, 255, 0), 2)
        cv.imshow(window_name, drawing_img)
        bgr = pixel.squeeze((0, 1))
        pixel = cv.cvtColor(pixel, cv.COLOR_BGR2HSV)
        hsv = pixel.squeeze((0, 1))
        print(f"bgr: {bgr}")
        print(f"hsv: {hsv}")
        print("#"*20)

cv.namedWindow(window_name)
cv.setMouseCallback(window_name, mouse_click)
cv.imshow(window_name, drawing_img)

if cv.waitKey() & 0xFF == 27:
    cv.destroyAllWindows()
```

```python
def pop_effect(img, lower, upper, debug=False):
    hsv = cv.cvtColor(img, cv.COLOR_BGR2HSV)
    fgr_mask = cv.inRange(hsv, lower, upper)
    bgr_mask = cv.bitwise_not(fgr_mask)
    bgr = cv.bitwise_and(img, img, mask=bgr_mask)
    fgr = cv.bitwise_and(img, img, mask=fgr_mask)
    bgr = cv.cvtColor(bgr, cv.COLOR_BGR2GRAY)
    bgr = cv.cvtColor(bgr, cv.COLOR_GRAY2BGR)
    final_img = cv.add(fgr, bgr)

    if debug:
        plt.figure(figsize=(4,16))
        plt.subplot(421),plt.imshow(img[...,::-1]);plt.title('img')
        plt.subplot(423),plt.imshow(fgr_mask, cmap="gray");plt.title('fgr_mask')
        plt.subplot(424),plt.imshow(bgr_mask, cmap="gray");plt.title('bgr_mask')
        plt.subplot(425),plt.imshow(fgr[...,::-1]);plt.title('fgr')
        plt.subplot(426),plt.imshow(bgr[...,::-1]);plt.title('bgr')
        plt.subplot(428),plt.imshow(final_img[...,::-1]);plt.title('final_img')

    return final_img
```

```python
window_name = "Pop Effects"
result_window_name = "Pop Effects Result"

def mouse_click(event, x, y, flags, param):
    if event == cv.EVENT_LBUTTONDOWN:
        drawing_img = input_img.copy()
        cv.circle(drawing_img, (x, y), 5, (0, 255, 0), 2)
        cv.imshow(window_name, drawing_img)
        pixel = input_img[y:y+1, x:x+1, :]
        pixel = cv.cvtColor(pixel, cv.COLOR_BGR2HSV)
        hsv = pixel.squeeze((0, 1))
        upper_h = min(179, int(hsv[0]) + 10)
        lower_h = max(0, int(hsv[0]) - 10)
        upper = np.array([upper_h, 255, 255], dtype=np.uint8)
        lower = np.array([lower_h, 50, 50], dtype=np.uint8)
        print(f"lower: {lower}")
        print(f"upper: {upper}")
        cv.imshow(result_window_name, pop_effect(input_img, lower, upper))


cv.namedWindow(window_name)
cv.setMouseCallback(window_name, mouse_click)
cv.imshow(window_name, input_img)

if cv.waitKey() & 0xFF == 27:
    cv.destroyAllWindows()
```
