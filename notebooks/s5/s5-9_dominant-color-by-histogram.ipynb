```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
bar = np.linspace(0, 179, 180, dtype="uint8")
bar = np.array([bar] * 40)
blank = np.ones_like(bar) * 255
bar = cv.merge([bar, blank, blank])
bar = cv.cvtColor(bar, cv.COLOR_HSV2BGR)
plt.imshow(bar[...,::-1])
plt.show()
```

```python
def plot_hsv_histo(img_src):
    img = cv.imread(img_src)
    hsv = cv.cvtColor(img, cv.COLOR_BGR2HSV)
    hist = cv.calcHist([hsv], [0], None, [180], [0, 180])
    plt.figure(figsize=(8,3))
    plt.subplot(121); plt.imshow(img[...,::-1])
    plt.subplot(222); plt.plot(hist)
    plt.subplot(224); plt.imshow(bar[...,::-1])
```

```python
plot_hsv_histo("../../images/3color.png")
```

```python
plot_hsv_histo("../../images/grayscale-pixels.png")
```

```python
plot_hsv_histo("../../images/man.jpg")
```
