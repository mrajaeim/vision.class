```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
bell_peper_img = cv.imread('../../../class.vision/images/felfel-dolme.jpg')
man_img = cv.imread('../../images/man.jpg')
```

```python
h_line = np.linspace(0, 180, 180, dtype=np.uint8)
h = np.array([h_line] * 30)
s = np.ones_like(h) * 255
v = np.ones_like(h) * 255
hsv_bar = cv.merge([h, s, v])
hsv_bar = cv.cvtColor(hsv_bar, cv.COLOR_HSV2BGR)
plt.imshow(hsv_bar[...,::-1])
```

```python
lower = np.array([20, 50, 50])
upper = np.array([35, 255, 255])

hsv_img = cv.cvtColor(bell_peper_img, cv.COLOR_BGR2HSV)
mask = cv.inRange(hsv_img, lower, upper)

output = cv.bitwise_and(bell_peper_img, bell_peper_img, mask=mask)

plt.figure(figsize = (10,10))
plt.subplot(131); plt.imshow(bell_peper_img[...,::-1])
plt.subplot(132); plt.imshow(mask, cmap = 'gray')
plt.subplot(133); plt.imshow(output[...,::-1])
```

```python
hsv_img = cv.cvtColor(bell_peper_img, cv.COLOR_BGR2HSV)

lower = np.array([0, 50, 50])
upper = np.array([10, 255, 255])
mask1 = cv.inRange(hsv_img, lower, upper)

lower = np.array([170, 50, 50])
upper = np.array([180, 255, 255])
mask2 = cv.inRange(hsv_img, lower, upper)

mask = cv.bitwise_or(mask1, mask2)

output = cv.bitwise_and(bell_peper_img, bell_peper_img, mask=mask)

plt.figure(figsize = (10,7))
plt.subplot(331); plt.imshow(bell_peper_img[...,::-1])
plt.subplot(334); plt.imshow(mask1, cmap = 'gray')
plt.subplot(335); plt.imshow(mask2, cmap = 'gray')
plt.subplot(336); plt.imshow(mask, cmap = 'gray')
plt.subplot(337); plt.imshow(output[...,::-1])

plt.tight_layout()
plt.show()
```

```python
hsv_img = cv.cvtColor(man_img, cv.COLOR_BGR2HSV)

lower = np.array([0, 50, 50])
upper = np.array([10, 255, 255])
mask1 = cv.inRange(hsv_img, lower, upper)

lower = np.array([170, 50, 50])
upper = np.array([180, 255, 255])
mask2 = cv.inRange(hsv_img, lower, upper)

mask = cv.bitwise_or(mask1, mask2)

output = cv.bitwise_and(man_img, man_img, mask=mask)

plt.figure(figsize = (10,7))
plt.subplot(331); plt.imshow(man_img[...,::-1])
plt.subplot(334); plt.imshow(mask1, cmap = 'gray')
plt.subplot(335); plt.imshow(mask2, cmap = 'gray')
plt.subplot(336); plt.imshow(mask, cmap = 'gray')
plt.subplot(337); plt.imshow(output[...,::-1])

plt.tight_layout()
plt.show()
```
