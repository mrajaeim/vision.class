```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

plt.style.use('default')
plt.style.use('seaborn-v0_8-whitegrid')
```

```python
gray_img = cv.imread('../../../class.vision/images/input.jpg',0)
dark_tones = cv.imread('../../../class.vision/images/dark.png')
mid_tones = cv.imread('../../../class.vision/images/mid.png')
light_tones = cv.imread('../../../class.vision/images/light.png')
```

```python
list_of_numbers = np.array([5,4,5,10,3,5,6,9,7,8,7,9,1,3,7])
plt.subplot(121); plt.hist(list_of_numbers)
plt.subplot(122); plt.hist(list_of_numbers, bins=3)
```

```python
plt.subplot(121); plt.hist(gray_img.ravel(), bins=256, color='orange')
plt.subplot(122); plt.hist(gray_img.ravel(), bins=256, histtype='step', color='red')
plt.show()
```

```python
hist = cv.calcHist([gray_img], [0], None, [256], [0,256])
plt.plot(hist)
```

```python
dark_tones_gray = cv.cvtColor(dark_tones, cv.COLOR_BGR2GRAY)
mid_tones_gray = cv.cvtColor(mid_tones, cv.COLOR_BGR2GRAY)
light_tones_gray = cv.cvtColor(light_tones, cv.COLOR_BGR2GRAY)

plt.figure(figsize = (10,6))
plt.subplot(231); plt.imshow(dark_tones_gray)
plt.subplot(234); plt.plot(cv.calcHist([dark_tones_gray], [0], None, [256], [0, 256]))
plt.subplot(232); plt.imshow(mid_tones_gray)
plt.subplot(235); plt.plot(cv.calcHist([mid_tones_gray], [0], None, [256], [0, 256]))
plt.subplot(233); plt.imshow(light_tones_gray)
plt.subplot(236); plt.plot(cv.calcHist([light_tones_gray], [0], None, [256], [0, 256]))
```

```python
channels = ('b','g','r')
images = [
    dark_tones,
    mid_tones,
    light_tones,
]
plt.figure(figsize = (10,6))
for idx_img, img in enumerate(images):
    plt.subplot(2,3, idx_img + 1); plt.imshow(img[...,::-1])
    plt.subplot(2,3, idx_img + 1 + 3)
    for idx_c, c in enumerate(channels):
        hist = cv.calcHist([img], [idx_c], None, [256], [0, 256])
        plt.plot(hist, color = c)
```
