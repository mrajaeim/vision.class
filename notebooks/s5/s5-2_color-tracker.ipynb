```python
from traceback import print_tb

import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
blue_cap_vid_pth = "../../../class.vision/videos/blue-track.mp4"
```

## Mask Blue Cap

```python
cap = cv.VideoCapture(blue_cap_vid_pth)
blue_range_lower = np.array([110, 50, 50])
blue_range_upper = np.array([130, 255, 255])

while 1:
    ret, frame = cap.read()
    if not ret:
        break

    hsv = cv.cvtColor(frame, cv.COLOR_BGR2HSV)
    mask = cv.inRange(hsv, blue_range_lower, blue_range_upper)
    masked_frame = cv.bitwise_and(frame, frame, mask=mask)

    cv.imshow('frame', frame)
    cv.imshow('mask', mask)
    cv.imshow('masked-frame', masked_frame)

    k = cv.waitKey(5) & 0xFF
    if k == 27:
        break

cv.destroyAllWindows()
cap.release()
```

## Mask Hand

```python
cap = cv.VideoCapture(blue_cap_vid_pth)

while 1:
    ret, frame = cap.read()
    if not ret:
        break

    hsv = cv.cvtColor(frame, cv.COLOR_BGR2HSV)

    lower = np.array([0, 50, 50])
    upper = np.array([10, 255, 255])
    mask1 = cv.inRange(hsv, lower, upper)

    lower = np.array([170, 50, 50])
    upper = np.array([180, 255, 255])
    mask2 = cv.inRange(hsv, lower, upper)

    mask = cv.bitwise_or(mask1, mask2)
    masked_frame = cv.bitwise_and(frame, frame, mask=mask)

    cv.imshow('frame', frame)
    cv.imshow('mask', mask)
    cv.imshow('masked-frame', masked_frame)

    k = cv.waitKey(5) & 0xFF
    if k == 27:
        break

cv.destroyAllWindows()
cap.release()
```

## Annotate object

```python
cap = cv.VideoCapture(blue_cap_vid_pth)
blue_range_lower = np.array([110, 50, 50])
blue_range_upper = np.array([130, 255, 255])

while 1:
    ret, frame = cap.read()
    if not ret:
        break

    hsv = cv.cvtColor(frame, cv.COLOR_BGR2HSV)
    mask = cv.inRange(hsv, blue_range_lower, blue_range_upper)

    contours, _ = cv.findContours(mask, cv.RETR_TREE, cv.CHAIN_APPROX_SIMPLE)
    center, radius = cv.minEnclosingCircle(contours[0])
    cv.circle(frame, (int(center[0]), int(center[1])), int(radius), color=(0, 0, 255), thickness=2)

    cv.imshow('frame', frame)

    k = cv.waitKey(5) & 0xFF
    if k == 27:
        break

cv.destroyAllWindows()
cap.release()
```
