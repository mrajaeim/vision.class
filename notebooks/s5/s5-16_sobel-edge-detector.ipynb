```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
little_girl_img = cv.imread('../../../class.vision/images/little_girl.jpg',0)
```

```python
grad_x = cv.Sobel(little_girl_img, cv.CV_32F, 1, 0, ksize=5)
grad_y = cv.Sobel(little_girl_img, cv.CV_32F, 0, 1, ksize=5)

final_sobel = cv.add(grad_x, grad_y)

plt.figure(figsize=[12,7])
plt.subplot(231);plt.imshow(little_girl_img, cmap='gray');plt.title("Original");
plt.subplot(234);plt.imshow(grad_x, cmap='gray');plt.title("sobel X");
plt.subplot(235);plt.imshow(grad_y, cmap='gray');plt.title("sobel Y");
plt.subplot(236);plt.imshow(final_sobel, cmap='gray');plt.title("Final");
```

```python
little_girl_img = cv.imread('../../../class.vision/images/little_girl.jpg')

little_girl_img = cv.GaussianBlur(little_girl_img, (5,5), 0)

gray = cv.cvtColor(little_girl_img, cv.COLOR_BGR2GRAY)

grad_x = cv.Sobel(gray, cv.CV_32F, 1, 0, ksize=3)
grad_y = cv.Sobel(gray, cv.CV_32F, 0, 1, ksize=3)

grad_x = cv.convertScaleAbs(grad_x)
grad_y = cv.convertScaleAbs(grad_y)

final_sobel = cv.addWeighted(grad_x, 0.5, grad_y, 0.5, 0)

plt.figure(figsize=[12,7])
plt.subplot(231);plt.imshow(little_girl_img[...,::-1]);plt.title("Original");
plt.subplot(234);plt.imshow(grad_x, cmap='gray');plt.title("sobel X");
plt.subplot(235);plt.imshow(grad_y, cmap='gray');plt.title("sobel Y");
plt.subplot(236);plt.imshow(final_sobel, cmap='gray');plt.title("Final");
```
