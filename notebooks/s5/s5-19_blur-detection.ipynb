```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
plate_cropped_img = cv.imread('../../../class.vision/images/plate_cropped.jpg',0)
plate3_cropped_img = cv.imread('../../../class.vision/images/plate3_cropped.jpg',0)
```

```python
def var_of_lag(img, ksize=3):
    l = cv.Laplacian(img, cv.CV_32F, ksize=ksize)
    return cv.convertScaleAbs(l), l.var()
```

```python
ksize = 5
l1, metric_image1 = var_of_lag(plate_cropped_img, ksize=ksize)
l2, metric_image2 = var_of_lag(plate3_cropped_img, ksize=ksize)


plt.figure(figsize=[12,4])
plt.subplot(221);plt.imshow(l1, cmap='gray')
plt.subplot(222);plt.imshow(l2, cmap='gray')
plt.subplot(223);plt.imshow(plate_cropped_img, cmap='gray');plt.title(metric_image1);
plt.subplot(224);plt.imshow(plate3_cropped_img, cmap='gray');plt.title(metric_image2);
plt.tight_layout()
diff = metric_image1 - metric_image2
print("metric1", ">" if diff > 0 else "<" ,"metric2")
print("diff 1-2:", diff)
```
