```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
def print_min_max(img):
    print("Max {}, Min {}, type {}".format(np.max(img), np.min(img), img.dtype))
```

```python
gradient_img = cv.imread('../../../class.vision/images/gradient.png',0)
sharp_variation_img = cv.imread('../../../class.vision/images/sharp_variation.png',0)
little_girl_img = cv.imread('../../../class.vision/images/little_girl.jpg',0)
```

```python
kernel = np.zeros((3,3), np.float32)
kernel[1,0] = -(1/2)
kernel[1,2] = (1/2)

print(kernel)

edges = cv.filter2D(gradient_img, -1, kernel)

plt.figure(figsize=[12,7])
plt.subplot(121);plt.imshow(gradient_img, cmap='gray');plt.title("Original");
plt.subplot(122);plt.imshow(edges, cmap='gray');plt.title("Edges");

print_min_max(gradient_img)
print_min_max(edges)
```

```python
kernel = np.zeros((3,3), np.float32)

kernel[1,0] = -(1/2)
kernel[1,2] = (1/2)
edges1 = cv.filter2D(little_girl_img, -1, kernel)

kernel[1,0] = (1/2)
kernel[1,2] = -(1/2)
edges2 = cv.filter2D(little_girl_img, -1, kernel)

all_vertical_edges = cv.add(edges1, edges2)

plt.figure(figsize=[15,7])
plt.subplot(231);plt.imshow(little_girl_img, cmap='gray');plt.title("Original");
plt.subplot(234);plt.imshow(edges1, cmap='gray');plt.title("Edges 1");
plt.subplot(235);plt.imshow(edges2, cmap='gray');plt.title("Edges 2");
plt.subplot(236);plt.imshow(all_vertical_edges, cmap='gray');plt.title("Final edges");
```

```python
kernel = np.zeros((3,3), np.float32)
kernel[1,0] = -(1/2)
kernel[1,2] = (1/2)

print(kernel)

edges = cv.filter2D(little_girl_img, cv.CV_64F, kernel)
edges = cv.convertScaleAbs(edges)

plt.figure(figsize=[12,7])
plt.subplot(121);plt.imshow(little_girl_img, cmap='gray');plt.title("Original");
plt.subplot(122);plt.imshow(edges, cmap='gray');plt.title("Edges");

print_min_max(little_girl_img)
print_min_max(edges)
```
