```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
plate_img = cv.imread('../../../class.vision/images/plate.jpg')
```

```python
k1 = np.zeros((3,3), np.float32)
k1[1,1] = 2
k2 = np.ones((3,3), np.float32) / 9
sharpening_kernel = k1 - k2
print(sharpening_kernel)

sharpened = cv.filter2D(plate_img, -1, sharpening_kernel)

plt.figure(figsize=[12,7])
plt.subplot(121);plt.imshow(plate_img[...,::-1]);plt.title("Original");
plt.subplot(122);plt.imshow(sharpened[...,::-1]);plt.title("Sharpened");
```

## More Sharper

```python
sharpening_kernel = np.full((3,3), -1, dtype=np.float32)
sharpening_kernel[1,1] = 11
print(sharpening_kernel)

sharpened = cv.filter2D(plate_img, -1, sharpening_kernel)

plt.figure(figsize=[12,7])
plt.subplot(121);plt.imshow(plate_img[...,::-1]);plt.title("Original");
plt.subplot(122);plt.imshow(sharpened[...,::-1]);plt.title("Sharpened");
```
